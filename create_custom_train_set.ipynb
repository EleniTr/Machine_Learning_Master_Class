##Create new train set

Clean dataset labels

```python
#Remove columns with highly repetative values
drop_cols = merged_train_set.columns[merged_train_set.nunique()==1]

#Remove survey_id,hhid,com
#Remove male as it creates a bias
drop_cols = list(drop_cols)+["survey_id","hhid","com","male"]

with open("drop_cols.txt", "w") as f:
    for col in drop_cols:
        f.write(col + "\n")

merged_train_set = merged_train_set.drop(columns=drop_cols)
```

One-Hot Encoding for categorial columns

```python
one_hot_cols = []
for col in merged_train_set.columns:
   #check for categorical strings
   if merged_train_set[col].dtype.name in ['object','category']:
    #remove NaN values and note down the values of the col
    unique_values = merged_train_set[col].dropna().unique()
    if len(unique_values)== 2:
      one_hot_cols.append(col)
print(one_hot_cols)

for col in one_hot_cols:
  unique_values = merged_train_set[col].dropna().unique()
  merged_train_set[col] = merged_train_set[col].map({unique_values[0]:0, unique_values[1]:1})
```

```python
merged_train_set
```

```python
# Find all non-numeric columns
non_numeric = merged_train_set.select_dtypes(exclude=['int64','float64']).columns
print("Non-numeric columns:", non_numeric)
```

```python
# #Extensive one-hot encoding for long categorical strings
# categorical_cols = merged_train_set.select_dtypes(include=['object','category']).columns
# long_categorical_cols = [col for col in categorical_cols if col not in one_hot_cols]
# print(long_categorical_cols)


base_names = []
for col in merged_train_set.columns:
  for base in non_numeric:
    if base in col and base not in base_names:
      base_names.append(base)

print(base_names)

# Find all columns that are the  base_name or contain it
columns_to_dummify = [col for col in merged_train_set.columns if any(base in col for base in base_names)]

print(columns_to_dummify)
merged_train_set = pd.get_dummies(merged_train_set,columns=columns_to_dummify,drop_first=True)

#Convert bool to binary
bool_cols = merged_train_set.select_dtypes(include=['bool']).columns
for col in bool_cols:
    merged_train_set[col] = merged_train_set[col].astype(int)

merged_train_set
```

```python
#Total food consumption
```

```python
total_consumtion = []
for col in merged_train_set.columns:
  if col.startswith("consumed"):
    total_consumtion.append(col)

merged_train_set["total_food_consumption"] = merged_train_set[total_consumtion].sum(axis=1)
merged_train_set = merged_train_set.drop(columns=total_consumtion)
```

###Mean number of children

```python
children = []
for col in merged_train_set.columns:
  if col.startswith("num_children"):
    children.append(col)

merged_train_set["mean_num_children"] = merged_train_set[children].sum(axis=1)
merged_train_set = merged_train_set.drop(columns=children)
```

Mean of working  adults

```python
employeed_adults = []
keywords = ["employed","sworkershh"]
for col in merged_train_set.columns:
  if any(k in col for k in keywords):
    employeed_adults.append(col)

merged_train_set["mean_employeed_adults"] = merged_train_set[employeed_adults].sum(axis=1)
merged_train_set = merged_train_set.drop(columns=employeed_adults)
```

```python
merged_train_set
```

Check for NaN values

```python
num_cols = merged_train_set.select_dtypes(include=['int64','float64']).columns
for col in num_cols:
  merged_train_set[col].fillna(merged_train_set[col].median(), inplace=True)

cat_cols = merged_train_set.select_dtypes(include=['object','category']).columns
for col in cat_cols:
  merged_train_set[col].fillna(merged_train_set[col].mode()[0], inplace=True)
```
